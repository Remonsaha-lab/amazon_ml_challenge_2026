# LambdaRank entity-resolution pipeline

This notebook calls `code_1/rank_pipeline.py`, which uses grouped LightGBM LambdaRank rather than the earlier XGBoost pairwise classifier.

The training section reports held-out validation metrics. The unlabeled test set can produce predictions but cannot produce a test score.


```python
from pathlib import Path
import sys
from argparse import Namespace

cwd = Path.cwd().resolve()
PROJECT_ROOT = next((p for p in (cwd, *cwd.parents) if (p / 'code_1' / 'rank_pipeline.py').is_file()), None)
if PROJECT_ROOT is None:
    raise FileNotFoundError('Open this notebook from the project checkout containing code_1/rank_pipeline.py.')
sys.path.insert(0, str(PROJECT_ROOT / 'code_1'))
from rank_pipeline import train, infer

print('Project root:', PROJECT_ROOT)
print('Dataset:', PROJECT_ROOT / 'dataset')
```

```python
# Tune these values before training if needed.
args = Namespace(
    train_cohort=30_000,
    validation_size=0.20,
    max_candidates=100,
    background_cap=400_000,
    n_estimators=600,
    seed=42,
    output=PROJECT_ROOT / 'output',
)
ranker, score_cutoff = train(args)
```

```python
# Run after training to create test predictions and candidate lists.
infer(ranker, score_cutoff, args)
```
